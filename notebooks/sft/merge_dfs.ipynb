```python
import pandas as pd
```

```python
df_1 = pd.read_csv("../../data/datasets/dpo.csv")
df_2 = pd.read_csv("../../data/datasets/la.csv")
df_3 = pd.read_csv("../../data/datasets/sft.csv")
df_4 = pd.read_csv("../../data/datasets/val.csv")
```

```python
# merge datasets
merged_df = pd.concat([df_1, df_2, df_3, df_4], ignore_index=True)
len(merged_df)
```

```python
# sample 500 rows as sft dataset and rest as grpo dataset
sft_df = merged_df.sample(n=500, random_state=42)
grpo_df = merged_df.drop(sft_df.index)
print(len(sft_df), len(grpo_df))
```

```python
sft_df.to_csv("../../data/datasets/sft.csv", index=False)
grpo_df.to_csv("../../data/datasets/grpo.csv", index=False)
```
