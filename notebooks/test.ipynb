```python
import json

def filter_taxonomy(input_path, output_path):
    processed_lines = 0
    kept_subjects = 0
    removed_subjects = 0

    with open(input_path, 'r', encoding='utf-8') as infile, \
         open(output_path, 'w', encoding='utf-8') as outfile:
        
        for line in infile:
            if not line.strip():
                continue
                
            try:
                data = json.loads(line)
                
                # Sprawdzamy czy linijka zawiera klucz 'subjects'
                if 'subjects' in data:
                    original_len = len(data['subjects'])
                    
                    # Zostawiamy tylko te, które NIE są "Foundational"
                    filtered_subjects = [
                        subject for subject in data['subjects'] 
                        if subject.get('level') != 'Foundational'
                    ]
                    
                    # Aktualizujemy listę w oryginalnym słowniku
                    data['subjects'] = filtered_subjects
                    
                    # Zliczanie statystyk
                    kept_subjects += len(filtered_subjects)
                    removed_subjects += (original_len - len(filtered_subjects))
                    
                    # Zapisujemy linijkę tylko, jeśli po filtrowaniu został jakikolwiek subject
                    if len(filtered_subjects) > 0:
                        json.dump(data, outfile, ensure_ascii=False)
                        outfile.write('\n')
                        processed_lines += 1
                        
            except json.JSONDecodeError as e:
                print(f"Błąd parsowania JSON w linijce: {e}")

    print("Zakończono filtrowanie!")
    print(f"Zapisane linie JSONL: {processed_lines}")
    print(f"Pozostawione (Advanced/Frontier) tematy: {kept_subjects}")
    print(f"Usunięte (Foundational) tematy: {removed_subjects}")
```

```python
input_filename = '../data/prompts/taxonomy_phase2_gemini-3-flash-preview.jsonl'
output_filename = '../data/prompts/taxonomy_phase2_gemini-3-flash-preview_adv.jsonl'

filter_taxonomy(input_filename, output_filename)
```
