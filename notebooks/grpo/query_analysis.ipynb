```python
import pandas as pd
```

```python
data = pd.read_csv("../../data/rl_grounded_dataset_merged.csv")
```

```python
data.head()
```

```python
num_papers = []
for row in data.itertuples():
    metadata = eval(row.metadata)
    num_papers.append(len(metadata["papers"]))
data["num_papers"] = num_papers
```

```python
data_filtered = data[(data["num_papers"] >= 3) & (data["num_papers"] <= 6)]
```

```python
data_filtered
```

```python
interestin_rows = [
    59, 70, 98, 114, 125, 129, 151, 227
]
interestin_rows_data = data_filtered.loc[interestin_rows]
interestin_rows_data
```

```python
interestin_rows_data.to_csv("../../data/interesting_rows.csv", index=False)
```

```python
human_frendly_df = pd.DataFrame(columns=["user_query", "hypothesis", "paper_1", "paper_2", "paper_3", "paper_4", "paper_5", "paper_6"])
for index, row in interestin_rows_data.iterrows():
    metadata = eval(row.metadata)
    papers = metadata["papers"]
    papers = [paper["summary"] for paper in papers]
    papers = papers[:6] + [""] * (6 - len(papers))  # Pad with empty strings if less than 6 papers
    human_frendly_df.loc[len(human_frendly_df)] = [row.user_query, row.hypothesis] + papers

human_frendly_df.to_csv("../../data/interesting_rows_human_friendly.csv", index=False)
```
